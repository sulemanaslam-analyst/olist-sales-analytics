```python
# Data quality check function
def data_quality_check(df, name):
    print(f"\n{'=' * 50}")
    print(f"DATA QUALITY CHECK: {name}")
    print(f"{'=' * 50}")

    # 1. Shape
    print("\n1. Dataset Shape:")
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    # 2. Column names
    print("\n2. Columns:")
    print(df.columns.tolist())

    # 3. Data types
    print("\n3. Data Types:")
    print(df.dtypes)

    # 4. Missing values
    print("\n4. Missing Values:")
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values.")
    else:
        print(missing)

    # 5. Duplicate rows
    print("\n5. Duplicate Rows:")
    print(df.duplicated().sum())

    # 6. Basic statistics
    print("\n6. Numerical Statistics:")
    print(df.describe())
```

```python
# Run the function
data_quality_check(orders, "Orders")
```

```python
# Data quality check function
def data_quality_check(df, name):

    print(f"\n{'=' * 60}")
    print(f"DATA QUALITY CHECK: {name}")
    print(f"{'=' * 60}")

    # 1. Dataset Shape
    print("\n1. Dataset Shape:")
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")

    # 2. First 5 Rows
    print("\n2. First 5 Rows:")
    print(df.head())

    # 3. Last 5 Rows
    print("\n3. Last 5 Rows:")
    print(df.tail())

    # 4. Column Names
    print("\n4. Columns:")
    print(df.columns.tolist())

    # 5. Data Types
    print("\n5. Data Types:")
    print(df.dtypes)

    # 6. Missing Values
    print("\n6. Missing Values:")
    missing = df.isnull().sum()
    missing = missing[missing > 0]

    if missing.empty:
        print("No missing values.")
    else:
        print(missing)

    # 7. Missing Value Percentage
    print("\n7. Missing Value Percentage:")
    missing_percent = (df.isnull().sum() / len(df)) * 100
    missing_percent = missing_percent[missing_percent > 0]

    if missing_percent.empty:
        print("No missing values.")
    else:
        print(missing_percent.round(2))

    # 8. Duplicate Rows
    print("\n8. Duplicate Rows:")
    print(df.duplicated().sum())

    # 9. Unique Values
    print("\n9. Unique Values:")
    print(df.nunique())

    # 10. Numerical Statistics
    print("\n10. Numerical Statistics:")
    print(df.describe())

    # 11. Categorical Statistics
    print("\n11. Categorical Statistics:")
    print(df.describe(include="object"))

    # 12. Order ID Uniqueness
    if "order_id" in df.columns:
        print("\n12. Order ID Check:")
        print(f"Total rows: {len(df)}")
        print(f"Unique order IDs: {df['order_id'].nunique()}")
        print(f"Duplicate order IDs: {df['order_id'].duplicated().sum()}")
```

```python
# Run the function
data_quality_check(customers, "Customers")
```

```python
# Run the function
data_quality_check(order_items, "Order_items")
```

```python
# Run the function
data_quality_check(products, "Products")
```
